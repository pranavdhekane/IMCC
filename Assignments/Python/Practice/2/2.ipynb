```python
orders = [
    {
        'id': 101,
        'customer': 'Asha',
        'status': 'DELIVERED',
        'items': [('keyboard', 2, 1499.0), ('mouse', 1, 599.0)]
    },
    {
        'id': 102,
        'customer': 'Ravi',
        'status': 'CANCELLED',
        'items': [('monitor', 1, 10999.0)]
    },
    {
        'id': 103,
        'customer': 'Asha',
        'status': 'DELIVERED',
        'items': [('cable', 5, 199.0)]
    },
    {
        'id': 104,
        'customer': 'Meera',
        'status': 'SHIPPED',
        'items': [('keyboard', 1, 1499.0), ('monitor', 1, 10999.0)]
    },
    {
        'id': 105,
        'customer': 'Ravi',
        'status': 'DELIVERED',
        'items': [('mouse', 3, 599.0)]
    }
]

def analyse(orders):
    revenue = dict()
    top_products = []
    all_products = {}
    all_customers = set()
    not_delivered = set()
    fully_delivered = []
    
    # Traversing through all orders..
    for order in orders:
        all_customers.add(order['customer'])
        
        if order['status'] == 'CANCELLED':
            not_delivered.add(order['customer'])
            continue
            
        if order['status'] == 'SHIPPED':
            not_delivered.add(order['customer'])
            
        revenuePerOrder = 0
        
        for item in order['items']:
            qty = item[1]
            val = item[2]
            revenuePerOrder += qty * val

            # Item is already present then add qty, otherwise set qty
            if item[0] in all_products:
                all_products[item[0]] += qty
            else:
                all_products[item[0]] = qty

        # If customer is already present add revenue else set revenue
        if order['customer'] in revenue:
            revenue[order['customer']] += revenuePerOrder
        else:
            revenue[order['customer']] = revenuePerOrder

    # Sorting top products
    top_products = sorted(all_products.items(), key=lambda p: (-p[1], p[0]))[:2]
    
    # fully delivered customers
    fully_delivered = sorted([customer for customer in all_customers if customer not in not_delivered])

    return {
        'revenue' : revenue, 'top_products': top_products, 'fully_delivered': fully_delivered
    }

print(analyse(orders))
```

```python
lines = [
    ' asha VERMA ; ASHA.V@Example.COM ; +91 98765-43210 ;pune ',
    'ravi kumar;ravi@@corp.in;9123456780;Mumbai',
    'meera;meera@mail;9123456780;Delhi',
    'john doe; john@x.org ;12345; Goa',
    'Neha Rao;neha.rao@firm.co.in;919876501234;nashik',
    'amit;amit@site.com;9876543210'
]

def clean_customer(raw):
    response = dict()
    
    details = raw.split(";")

    if len(details) != 4:
        return None

    details = [i.strip() for i in details]

    # Name check
    if not len(details[0]):                              
        return None                                      
    for word in details[0].split():                     
        if not word.isalpha():                           
            return None                                 

    # Empty city check
    if not len(details[3]):
        return None

    # email cleaning 
    mail = details[1].lower()
    if mail.count('@') != 1:
        return None

    pos = mail.find('@')
    if pos == 0 or pos == len(mail) - 1:
        return None

    domain = mail[pos+1:]
    if '.' not in domain:
        return None
    if domain.startswith('.') or domain.endswith('.'):
        return None

    # phone number cleaning
    phone = details[2]
    phone = phone.replace(' ', '').replace('-', '')      
    if phone.startswith('+91'):                          
        phone = phone[3:]                                
    elif len(phone) == 12 and phone.startswith('91'):   
        phone = phone[2:]                                

    if len(phone) != 10 or not phone.isdigit():          
        return None

    if phone[0] not in '6789':
        return None
    
    response['name'] = details[0].title()
    response['email'] = mail                             
    response['phone'] = phone
    response['city'] = details[3].title()
    
    return response

from pprint import pprint
invalid_count = 0
for i in lines:
    res = clean_customer(i)
    if res is None:
        invalid_count +=1
    else:
        pprint(res, sort_dicts=False)                   
print(f"Rejected: {invalid_count}")
```
